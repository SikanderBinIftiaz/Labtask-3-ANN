```python
import numpy as np
import pandas as pd
df=pd.read_csv("Obesity_Classification.csv")
df.head()
```

```python
df = df.drop(columns=["ID"]).dropna()
df["Gender"] = np.where(df["Gender"] == "Male", 1, 0)
df["Target"] = np.where(df["Label"].isin(["Overweight", "Obese"]), 1, 0)
print("Class counts -> 0 (Normal/Underweight):", int((df["Target"] == 0).sum()), "| 1 (Overweight/Obese):", int((df["Target"] == 1).sum()))
X = df[["Age", "Gender", "Height", "Weight", "BMI"]].values.astype(float)
y = df["Target"].values
np.random.seed(42)
idx = np.random.permutation(len(X))
X, y = X[idx], y[idx]
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
std[std == 0] = 1.0
X_train = (X_train - mean) / std
X_test = (X_test - mean) / std
weights = np.zeros(X_train.shape[1])
bias = 0.0
lr = 0.01
epochs = 50
for epoch in range(epochs):
    for xi, target in zip(X_train, y_train):
        prediction = 1 if np.dot(xi, weights) + bias >= 0 else 0
        error = target - prediction
        weights = weights + lr * error * xi
        bias = bias + lr * error
def predict(X):
    return np.where(np.dot(X, weights) + bias >= 0, 1, 0)
train_acc = float(np.mean(predict(X_train) == y_train))
y_pred = predict(X_test)
tp = int(np.sum((y_test == 1) & (y_pred == 1)))
tn = int(np.sum((y_test == 0) & (y_pred == 0)))
fp = int(np.sum((y_test == 0) & (y_pred == 1)))
fn = int(np.sum((y_test == 1) & (y_pred == 0)))
accuracy = (tp + tn) / len(y_test)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0
f1 = 2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0.0
print("Training Accuracy:", round(train_acc, 4))
print("Test Accuracy    :", round(accuracy, 4))
print("Precision        :", round(precision, 4))
print("Recall           :", round(recall, 4))
print("F1-Score         :", round(f1, 4))
print("Confusion Matrix:")
print([[tn, fp], [fn, tp]])
```
